## 🔒 Part 2 — Require Pull Request Reviews
You must require reviews before merging.

1. Go to **Settings → Branches**
2. Click **Add branch protection rule**
3. Apply the rule to `main`
4. Enable:
- ✅ Require a pull request before merging
- ✅ Require approvals
- Set required approvals to **1**
- ✅ (Recommended) Require conversation resolution before merging
5. Save the rule

### ⚠️ Copilot Note
- **Do NOT add Copilot as a reviewer**
- If you *do* add Copilot anyway, set **required approvals = 2**
so both **Copilot + your buddy** must approve.

## 🏓 Part 3 — The Ping-Pong Workflow (Minimum Required)
You must complete **at least 5 back-and-forth interactions**.

### Rules
- ❌ Do **not** push directly to `main`
- ✅ All work must be: `branch → pull request → review → merge → delete branch`
- Both partners must create branches and PRs

### What counts as an interaction?
- Partner A opens a PR → Partner B reviews → merge  
- Partner B opens a PR → Partner A reviews → merge  

Repeat until you have **at least 5 total ping-pong interactions**.

📊 The grader will inspect your **GitHub Network graph** to verify this.
